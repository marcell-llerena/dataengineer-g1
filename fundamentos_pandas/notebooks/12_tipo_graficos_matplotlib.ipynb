### Gráfico de lineas

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0, 5, 80)
y = np.random.random(len(x))

plt.plot(x, y, marker="o", linestyle="--", color="b")
plt.xlabel("Periodo")
plt.ylabel("Valor")
plt.title("Gráfico de líneas")
plt.show()
```

### Gráfico de barras

```python
categorias = ["A", "B", "C", "D"]
valores = np.random.randint(0, 20, size=len(categorias))

plt.bar(categorias, valores, label=categorias, color=["red", "blue", "green", "orange"], width=0.5)
plt.xlabel("Categoria")
plt.ylabel("Frecuencia")
plt.title("Gráfico de barras")
plt.legend(title="Categorias")
plt.show()
```

```python
plt.barh(categorias, valores, label=categorias, color=["red", "blue", "green", "orange"])
plt.xlabel("Frecuencia")
plt.ylabel("Categoria")
plt.title("Gráfico de barras")
plt.legend(title="Categorias")
plt.show()
```

### Gráfico de pastel

```python
frutas = ["Manzanas", "Peras", "Granadillas", "Fresas"]
porcentajes = np.random.randint(0, 100, len(frutas))

plt.pie(porcentajes, labels=frutas, autopct="%1.1f%%")
plt.title("Gráfico de pastel")
plt.show()
```

### Gráfico de dispersión

```python
x = np.random.rand(250)
y = np.random.rand(250)

plt.scatter(x, y, color="blue")
plt.xlabel("Variable (x)")
plt.ylabel("Variable (y)")
plt.title("Gráfico de dispersión")
plt.show()
```
