```python
import numpy as np
import matplotlib

print(matplotlib.__version__)
print(np.__version__)
```

```python
import matplotlib.pyplot as plt

x = np.linspace(0, 5, 10)
y = x ** 3

plt.plot(x, y)
plt.show()
```

```python
plt.plot(x, y, "r")
plt.show()
```

```python
plt.plot(x, y, "ro")
plt.show()
```

```python
plt.plot(x, y, "go-")
plt.show()
```

```python
plt.plot(x, y, "gD:")
plt.show()
```

```python
plt.plot(x, y, color="#B900FF", linestyle="--", marker="o")
```

```python
# https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html
# https://htmlcolorcodes.com/es/

plt.plot(x, y, color="#00F4FF", linestyle="--", linewidth= 2.5, marker="D")
```
