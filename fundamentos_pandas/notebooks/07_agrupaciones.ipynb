```python
import pandas as pd

ventas_df = pd.read_csv(
    "../data/ventas.csv"
)

ventas_df["importe"] = ventas_df["unidades"] * ventas_df["precio_unitario"]
ventas_df.head(5)
```

```python
# Forma permitida pero muy lenta
ventas_df[ventas_df["region"] == "Sur"]["importe"].sum()
```

```python
ventas_df.groupby("region")["importe"].sum()
```

```python
ventas_df[ventas_df["region"] == "Sur"]["importe"].mean()
```

```python
ventas_df.groupby("region")["importe"].mean()
```

```python
ventas_df["region"].value_counts()
```

#### Adicionando más stats a mi función agregada

```python
ventas_df.groupby("region")["importe"].agg(["sum", "mean", "median", "count"])
```

#### Agregando alias a mis operaciones agregadas

```python
ventas_df.groupby("region")["importe"].agg(
    importe_total="sum",
    importe_promedio="mean",
    importe_mediana="median",
    cantidad_registros="count"
)
```

#### Agrupar por n-columnas

```python
ventas_df.groupby(["region", "categoria"])["importe"].agg(
    importe_total="sum",
    importe_promedio="mean",
    total_registros="count"
)
```

```python
ventas_df.groupby("region").agg({
    "importe": ["sum", "mean", "max"],
    "unidades": ["sum"],
    "producto": ["nunique"]
})
```

```python
ventas_df.groupby(["region", "categoria", "vendedor"]).agg({
    "importe": ["sum", "mean", "max"],
    "unidades": ["sum"],
    "producto": ["nunique"]
})
```

```python
ventas_df.groupby("fecha")["importe"].agg(["sum", "count"])
```
