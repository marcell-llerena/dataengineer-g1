### Concatenado de tablas

```python
import pandas as pd

enero = pd.DataFrame({
    "mes": ["Enero"],
    "numero_mes": [1]
})

febrero = pd.DataFrame({
    "mes": ["Febrero"],
    "numero_mes": [2]
})

marzo = pd.DataFrame({
    "mes": ["Marzo"],
    "numero_mes": [3]
})
```

```python
febrero
```

```python
### Concatenar por columnas
pd.concat([enero, febrero, marzo], axis=1)
```

```python
### Concatenar por filas
pd.concat([enero, febrero, marzo], axis=0)
```

```python
pd.concat([enero, febrero, marzo], axis=0).reset_index()
```

```python
### Concatenar por filas y resetear índices
pd.concat([enero, febrero, marzo], axis=0).reset_index(drop=True)
```

### JOINS

```python
ventas_df = pd.read_csv("../data/ventas.csv")
catalogo_productos_df = pd.read_csv("../data/catalogo_productos.csv")
```

```python
ventas_df.head(5)
```

```python
catalogo_productos_df.head(10)
```

#### LEFT JOIN

```python
ventas_df.shape
```

```python
left_join = ventas_df.merge(catalogo_productos_df, on="producto", how="left")
left_join
```

```python
ventas_df.loc[1199, "producto"] = "Monitor 32"
ventas_df.tail()
```

```python
left_join = ventas_df.merge(catalogo_productos_df, on="producto", how="left")
left_join
```

### INNER JOIN

```python
inner_join = ventas_df.merge(catalogo_productos_df, on="producto", how="inner")
inner_join
```

### RIGHT JOIN

```python
nueva_fila = pd.DataFrame([{
    "producto": "iPhone 18 Pro Max",
    "coste_unitario": 1000,
    "proveedor": "Apple",
    "stock": 250
}])
nueva_fila
```

```python
catalogo_productos_df = pd.concat([catalogo_productos_df, nueva_fila], axis=0).reset_index(drop=True)
catalogo_productos_df
```

```python
right_join = ventas_df.merge(catalogo_productos_df, on="producto", how="right")
right_join
```

### LEFT ANTI JOIN

```python
left_anti_join = ventas_df.merge(catalogo_productos_df, on="producto", how="left_anti")
left_anti_join
```

### RIGHT ANTI JOIN

```python
right_anti_join = ventas_df.merge(catalogo_productos_df, on="producto", how="right_anti")
right_anti_join
```

### CROSS JOIN

```python
cross_join = catalogo_productos_df.merge(catalogo_productos_df, how="cross")
cross_join.head(50)
```
