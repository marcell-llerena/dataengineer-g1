### CTE - COMMON TABLE EXPRESSIONS

```sql
USE SCHEMA tienda;
```

```sql
WITH total_historico_por_cliente AS (
    SELECT 
        cliente_id,
        SUM(total) AS total_historico
    FROM pedidos
    GROUP BY cliente_id
)
SELECT
    th.*,
    cl.nombre
FROM total_historico_por_cliente  AS th
INNER JOIN clientes AS cl
ON th.cliente_id = cl.id
ORDER BY th.total_historico
```

```sql
SELECT
    th.*,
    cl.nombre
FROM (
    SELECT
        cliente_id,
        SUM(total) AS total_historico
    FROM pedidos
    GROUP BY cliente_id
) AS th
INNER JOIN clientes AS cl
ON th.cliente_id = cl.id
ORDER BY th.total_historico DESC
```
