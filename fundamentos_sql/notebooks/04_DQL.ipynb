```sql
SELECT *
FROM productos;
```

```sql
SELECT nombre, precio
FROM productos;
```

```sql
SELECT  
    nombre AS nombre_apellidos,
    pais AS codigo_pais,
    telefono AS telefono_principal
FROM clientes;
```

```sql
SELECT 
    nombre,
    precio,
    stock,
    precio * stock AS precio_total_stock
FROM productos;
```

```sql
SELECT *
FROM productos
WHERE stock > 100;
```

```sql
SELECT *
FROM productos
WHERE stock > 100 AND precio > 30;
```

```sql
SELECT *
FROM productos
WHERE precio BETWEEN 10 AND 100;
```

```sql
SELECT *
FROM productos
WHERE precio BETWEEN 10 AND 100
ORDER BY id;
```

```sql
SELECT *
FROM clientes
WHERE pais = "ES" or pais = "MX"
```

```sql
SELECT *
FROM clientes
WHERE pais IN ("ES", "MX")
```

```sql
SELECT *
FROM clientes
WHERE nombre LIKE '%ez';
```

```sql
SELECT *
FROM clientes
WHERE telefono LIKE '6%';
```

```sql
SELECT
    nombre,
    UPPER(nombre) AS nombre_mayusculas,
    LOWER(nombre) AS nombre_minusculas,
    TRIM(nombre) AS nombre_sin_espacios,
    pais,
    CASE
        WHEN pais == "ES" THEN "España"
        WHEN pais == "MX" THEN "México"
        WHEN pais == "FR" THEN "Francia"
        ELSE "Otro"
    END AS nombre_pais
FROM clientes;
```

```sql
SELECT *
FROM clientes
WHERE pais != "MX"; -- <>
```

```sql
SELECT
    nombre,
    pais,
    telefono,
    COALESCE(telefono, "NO TELEFONO") AS telefono_tratado
FROM clientes
WHERE pais != "MX"; -- <>
```

```sql
UPDATE clientes
SET nombre = "ana lópez"
WHERE id = 1;
```

```sql
SELECT
    nombre,
    INITCAP(nombre) nombre_modo_titulo
FROM clientes
```

```sql
SELECT
    nombre,
    pais,
    CONCAT(nombre, " : ", pais) AS concatenado
FROM clientes
```

```sql
SELECT
    nombre AS nombre_completo,
    SPLIT(nombre, " ")[0] AS nombre,
    SPLIT(nombre, " ")[1] AS apellido
FROM clientes
```

```sql
SELECT
    telefono,
    LENGTH(telefono) AS longitud,
    RPAD(telefono, 11, "0") AS telefono_tratado,
    LPAD(telefono, 11, "0") AS telefono_tratado
FROM clientes;
```

```sql
SELECT
    id,
    POW(id, 2) AS power_2,
    POW(id, 3) AS power_3,
    SQRT(id) AS raiz_2
FROM clientes;
```
