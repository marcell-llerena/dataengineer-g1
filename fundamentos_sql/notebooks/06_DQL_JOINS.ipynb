### LEFT JOIN

```sql
SELECT *
FROM clientes;
```

```sql
SELECT
    pe.*,
    cl.nombre,
    cl.telefono
FROM pedidos AS pe
LEFT JOIN clientes AS cl
ON pe.cliente_id = cl.id
ORDER BY cliente_id, fecha;
```

### INNER JOIN

```sql
SELECT
    pe.*,
    cl.nombre,
    cl.telefono
FROM pedidos AS pe
INNER JOIN clientes AS cl
ON pe.cliente_id = cl.id
ORDER BY cliente_id, fecha;
```

### LEFT ANTI JOIN

```sql
SELECT
    pe.*
FROM pedidos AS pe
LEFT ANTI JOIN clientes AS cl
ON pe.cliente_id = cl.id
ORDER BY cliente_id, fecha;
```

### FULL OUTER JOIN

```sql
SELECT
    pe.*,
    cl.*
FROM pedidos AS pe
FULL OUTER JOIN clientes AS cl
ON pe.cliente_id = cl.id
ORDER BY cliente_id, fecha;
```

```sql
SELECT *
FROM pedidos;
```

```sql
SELECT
    pe.id AS pedido_id,
    pe.fecha AS fecha_pedido,
    cli.nombre AS nombre_cliente,
    dp.cantidad AS cantidad_producto,
    pr.nombre AS nombre_producto
FROM pedidos AS pe
INNER JOIN clientes AS cli
ON pe.cliente_id = cli.id
INNER JOIN detalle_pedidos AS dp
ON pe.id = dp.pedido_id
INNER JOIN productos AS pr
ON dp.producto_id = pr.id
ORDER BY pe.cliente_id, pe.fecha;
```

```sql
CREATE OR REPLACE VIEW vw_informe_detallado_pedidos AS
    SELECT
        pe.id AS pedido_id,
        pe.fecha AS fecha_pedido,
        cli.nombre AS nombre_cliente,
        dp.cantidad AS cantidad_producto,
        pr.nombre AS nombre_producto
    FROM pedidos AS pe
    INNER JOIN clientes AS cli
    ON pe.cliente_id = cli.id
    INNER JOIN detalle_pedidos AS dp
    ON pe.id = dp.pedido_id
    INNER JOIN productos AS pr
    ON dp.producto_id = pr.id
    ORDER BY pe.cliente_id, pe.fecha;
```

```sql
SELECT COUNT(1)
FROM vw_informe_detallado_pedidos;
```

```sql
CREATE OR REPLACE TABLE tb_informe_detallado_pedidos AS
    SELECT
        pe.id AS pedido_id,
        pe.fecha AS fecha_pedido,
        cli.nombre AS nombre_cliente,
        dp.cantidad AS cantidad_producto,
        pr.nombre AS nombre_producto
    FROM pedidos AS pe
    INNER JOIN clientes AS cli
    ON pe.cliente_id = cli.id
    INNER JOIN detalle_pedidos AS dp
    ON pe.id = dp.pedido_id
    INNER JOIN productos AS pr
    ON dp.producto_id = pr.id
    ORDER BY pe.cliente_id, pe.fecha;
```

```sql
SELECT COUNT(1)
FROM tb_informe_detallado_pedidos;
```
