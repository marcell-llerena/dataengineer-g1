```sql
SELECT
    ca.nombre AS categoria,
    COUNT(1) AS total_productos,
    SUM(pr.precio) AS total_recaudado,
    AVG(pr.precio) AS precio_promedio,
    MAX(pr.precio) AS precio_maximo,
    MIN(pr.precio) AS precio_minimo,
    SUM(pr.stock) AS total_stock
FROM productos AS pr
INNER JOIN categorias AS ca
ON pr.categoria_id = ca.id
GROUP BY ca.nombre
```
