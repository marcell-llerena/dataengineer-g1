# AGREGACIONES
- COUNT
- AVG
- MIN
- MAX
- SUM

```sql
USE SCHEMA tienda;
```

```sql
SELECT *
FROM productos;
```

```sql
SELECT
    COUNT(1) AS total_registros,
    MAX(precio) AS precio_maximo,
    MIN(precio) AS precio_minimo,
    AVG(precio) AS precio_promedio,
    SUM(precio) AS precio_total,
    MEDIAN(precio) AS precio_medio,
    MODE(precio) AS precio_moda
FROM productos;
```
