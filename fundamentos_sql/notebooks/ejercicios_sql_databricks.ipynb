## 3. Tablas y esquema

| Tabla | Columnas y tipos recomendados | Descripción |
|---|---|---|
| `categorias` | `category_id INT`; `category_name STRING` | Catálogo de categorías. |

| `productos` | `product_id INT`; `product_name STRING`; `category_id INT`; `unit_price DECIMAL(10,2)`; `stock INT`; `active BOOLEAN` | Catálogo de productos. |
| `pedidos` | `order_id INT`; `customer_id INT`; `employee_id INT`; `order_date DATE`; `status STRING`; `channel STRING` | Cabecera de pedidos. |
