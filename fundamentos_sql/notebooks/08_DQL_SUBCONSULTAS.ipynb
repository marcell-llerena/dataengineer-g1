```sql
USE SCHEMA tienda;
```

```sql
SELECT AVG(precio) AS precio_promedio
FROM productos;
```

```sql
SELECT *
FROM productos
WHERE precio > 180.3108333;
```

```sql
SELECT *
FROM productos
WHERE precio > (
    SELECT AVG(precio) AS precio_promedio
    FROM productos
);
```

```sql
SELECT *
FROM pedidos
```

```sql
SELECT *
FROM clientes
WHERE id NOT IN (
    SELECT cliente_id
    FROM pedidos
);
```

```sql
SELECT *
FROM (
    SELECT categoria_id, AVG(precio) promedio_por_categoria
    FROM productos
    GROUP BY categoria_id
)
WHERE promedio_por_categoria > 100;
```
