```sql
-- ESPECIFICAR EL CATALOGO
CREATE TABLE IF NOT EXISTS workspace.tienda.categorias (
    id      INT     NOT NULL,
    nombre  STRING  NOT NULL
);

-- NO ESPECIFICARLO PERO POR DEFECTO UTILIZA "workspace"
CREATE TABLE IF NOT EXISTS tienda.productos (
    id INT NOT NULL,
    nombre STRING NOT NULL,
    precio DECIMAL(10, 3),
    stock INT,
    categoria_id INT
);

CREATE TABLE IF NOT EXISTS tienda.detalle_pedidos (
    pedido_id INT NOT NULL,
    producto_id INT NOT NULL,
    cantidad INT,
    precio_unit DECIMAL(10, 3)
);

CREATE TABLE IF NOT EXISTS tienda.pedidos (
    id INT NOT NULL,
    cliente_id INT NOT NULL,
    fecha DATE,
    total DECIMAL(10, 3)
);

CREATE TABLE IF NOT EXISTS tienda.clientes (
    id INT NOT NULL,
    nombre STRING,
    email STRING,
    pais STRING,
    telefono STRING
);
```
