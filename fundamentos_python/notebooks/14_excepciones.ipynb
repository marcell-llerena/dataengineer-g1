```python
fila = {
    "id": "V101",
    "precio": "200.50",
    "cantidad": "100"
}
```

```python
try:
    precio_total = fila["precio"] + 50
except TypeError:
    try:
        precio_total = int(fila["precio"]) + 50
        print(f"cantidad total {precio_total}")
    except ValueError:
        precio_total = float(fila["precio"]) + 50
        print(f"cantidad total {precio_total}")
```

```python
try:
    precio_total = fila["precio"] + 50
except (ValueError, TypeError) as e:
    precio_total = float(fila["precio"]) + 50
    print(f"El error generado es: {e}")
    print(f"cantidad total {precio_total}")
```
