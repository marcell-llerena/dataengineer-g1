```python
regiones = ["Lima", "Cuzco", "Lima", "Arequipa", "Cuzco", "Piura"]

set_regiones = set(regiones)

print(f"lista regiones: {regiones}")
print(f"conjunto (set) regiones: {set_regiones}")
```

```python
len(set_regiones)
```

```python
sedes_esperadas = {"Arequipa", "Piura", "Loreto"}
```

### Unión ("|" - or)

```python
set_regiones
```

```python
sedes_esperadas
```

```python
set_regiones | sedes_esperadas
```

```python
set_regiones.union(sedes_esperadas)
```

### Intersección ("&" - and)

```python
set_regiones
```

```python
sedes_esperadas
```

```python
set_regiones & sedes_esperadas
```

```python
set_regiones.intersection(sedes_esperadas)
```

### Diferencia ("-")

```python
set_regiones
```

```python
set_regiones - sedes_esperadas
```

```python
sedes_esperadas - set_regiones
```

```python
sedes_esperadas.difference(set_regiones)
```

### Diferencia simétrica ("^")

```python
set_regiones
```

```python
sedes_esperadas
```

```python
set_regiones ^ sedes_esperadas
```

```python
set_regiones.symmetric_difference(sedes_esperadas)
```
