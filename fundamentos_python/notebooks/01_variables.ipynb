### TIPO DE VARIABLES

```python
cantidad: int = 3
precio: float = 145.0
cliente: str = "Marco Torres"
activo: bool = True
descuento: None = None
```

```python
print("Tipo de la variable cantidad: ", type(cantidad))
print("Tipo de la variable precio: ", type(precio))
print("Tipo de la variable descuento: ", type(descuento))
```

### CASTEO DE VARIABLES

```python
cantidad = "3"
print(type(cantidad))

cantidad = int(cantidad)
print(type(cantidad))
```

```python
precio = "130.5"
print(type(precio))

precio = float(precio)
print(type(precio))
print(precio)

precio = int(precio)
print(precio)
```

```python
activo = 0
print(activo)
activo = bool(activo)
print(activo)
```

```python
activo = -1000
print(activo)

activo = bool(activo)
print(activo)
```

```python
activo = ""
activo = bool(activo)
print(activo)
```

```python
activo = "False"
activo = bool(activo)
print(activo)
```
