### Programación Orientada a Objetcos (POO)

```python
class Perro:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad

    def ladrar(self):
        return f"{self.nombre} dice: Guau!"

    def cumplir_anios(self):
        self.edad += 1
```

```python
fido = Perro("Fido", 5)
```

```python
print(fido.ladrar())
```

```python
print(fido.edad)
```

```python
fido.cumplir_anios()
```

```python
print(fido.edad)
```

```python
toby = Perro("Toby", 10)
toby.ladrar()
```

```python
class Rectangulo:
    def __init__(self, base: int | float, altura: int | float) -> None:
        self.base = base
        self.altura = altura

    def area(self) -> int | float:
        return self.base * self.altura

    def perimetro(self) -> int | float:
        return self.base * 2 + self.altura * 2
```

```python
rectangulo = Rectangulo(10, 3)

print(f"Area: {rectangulo.area()}")
print(f"Perímetro: {rectangulo.perimetro()}")
```

```python
rectangulo_2 = Rectangulo(5, 4)

print(f"Area: {rectangulo_2.area()}")
print(f"Perímetro: {rectangulo_2.perimetro()}")
```

### Encapsulamiento

```python
class CuentaBancaria:
    def __init__(self, titular: str, saldo: float = 0.00) -> None:
        self.titular = titular
        self._saldo = saldo

    def _monto_positivo(self, monto: float) -> bool:
        return monto > 0

    def depositar(self, monto: float) -> None:
        if self._monto_positivo(monto):
            self._saldo += monto

    def retirar(self, monto: float) -> None:
        if self._monto_positivo(monto):
            self._saldo -= monto

    def ver_saldo(self) -> float:
        return self._saldo
```

```python
cuenta_1 = CuentaBancaria("Ana", 100)
print(f"saldo inicial: {cuenta_1.ver_saldo()}")

cuenta_1.depositar(50)
print(f"saldo luego de depositar: {cuenta_1.ver_saldo()}")


cuenta_1.retirar(20)
print(f"saldo luego de retirar: {cuenta_1.ver_saldo()}")
```

### HERENCIA

```python
class Animal:
    def __init__(self, nombre: str):
        self.nombre = nombre

    def presentarse(self) -> str:
        return f"Soy {self.nombre}"

### MALA PRACTICA
class Perro(Animal):
    def ladrar(self) -> str:
        return "Guau!"

### BUENA PRACTICA
class Perro(Animal):
    def __init__(self, nombre: str) -> None:
        super().__init__(nombre)

    def ladrar(self) -> str:
        return "Guau!"

class Gato(Animal):
    def __init__(self, nombre: str, color: str) -> None:
        super().__init__(nombre)
        self.color = color
```

```python
toby = Perro("Toby")
print(toby.ladrar())
```

```python
print(toby.presentarse())
```

```python
lucifer = Gato("Lucifer", "negro")
print(lucifer.presentarse())
```
