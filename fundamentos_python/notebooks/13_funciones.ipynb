```python
def suma(numero_1: float, numero_2: float) -> float:
    return numero_1 + numero_2
```

```python
suma(20, 30)
```

```python
suma(50, 80)
```

```python
PI = 3.1416

def calc_area_circulo(radio: float, n_decimales: int = 2) -> float:
    return round(PI * radio**2, n_decimales)
```

```python
calc_area_circulo(10, 4)
```

```python
calc_area_circulo(3, 3)
```

```python
calc_area_circulo(3)
```
