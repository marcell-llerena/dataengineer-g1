```python
precios = [2300, 5000, 9000]

con_igv = []

for precio in precios:
    con_igv.append(precio * 1.18)

print(con_igv)
```

### LISTA POR COMPRENSIÓN

```python
precios = [2300, 5000, 9000]

con_igv = [precio * 1.18 for precio in precios]

print(con_igv)
```

```python
ventas = [
    {"region": "Arequipa", "monto": 6400},
    {"region": "Puno", "monto": 0.0},
    {"region": "Lima", "monto": 5.0},
    {"region": "Tacna", "monto": 0.0},
]

validas = []

for venta in ventas:
    if venta["monto"] > 0:
        validas.append(venta)

print(validas)
```

```python
validas = [venta for venta in ventas if venta["monto"] > 0]

print(validas)
```
